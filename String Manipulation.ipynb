```python
# 3️⃣ String Manipulation in Python (for Bioinformatics)
# 3.1 Creating Strings
dna = "ATGCGTACGTTAGC"
rna = "AUGCGUACGUUAGC"
protein = "MKTLL"

# Multiline strings:

sequence = """ATGCGT
ACGTTAGC"""
print(sequence)
# 3.2 Accessing Characters and Slices
dna = "ATGCGTACGTTAGC"

print(dna[0])       # A → first base
print(dna[-1])      # C → last base
print(dna[0:3])     # ATG → first 3 bases (slicing)
print(dna[3:])      # CGTACGTTAGC → from 4th base to end
print(dna[:6])      # ATGCGT → first 6 bases
# 3.3 String Length
print(len(dna))  # 14 → sequence length
# 3.4 Basic String Methods
dna = "atgcgtacgttagc"

print(dna.upper())  # ATGCGTACGTTAGC
print(dna.lower())  # atgcgtacgttagc
print(dna.count("A"))  # Count A bases
print(dna.replace("T", "U"))  # Convert DNA → RNA
# 3.5 Checking for Motifs or Substrings
dna = "ATGCGTACGTTAGC"

print("ATG" in dna)   # True → start codon found
print("TTT" in dna)   # False
# 3.6 Finding Positions (indexing)
dna = "ATGCGTACGTTAGC"

print(dna.find("ATG"))  # 0 (start at index 0)
print(dna.find("TAA"))  # -1 (not found)
# 3.7 Reverse Complement (common in bioinformatics)
dna = "ATGCGTACGTTAGC"

# Reverse
rev = dna[::-1]

# Complement
complement = dna.replace("A", "t").replace("T", "a").replace("G", "c").replace("C", "g").upper()

# Reverse complement
rev_complement = complement[::-1]

print("DNA:", dna)
print("Reverse Complement:", rev_complement)
# 3.8 Splitting and Joining Strings
genes = "BRCA1,TP53,EGFR,MYC"
gene_list = genes.split(",")   # Split into list
print(gene_list)

joined_genes = "-".join(gene_list)
print(joined_genes)
# 3.9 Removing Whitespace or Newlines
seq = " ATGCGTACGTTAGC\n"
print(seq.strip())  # Removes spaces/newlines
# 3.10 Formatting Strings (f-string)
gene = "TP53"
gc = 0.48
length = 1200

print(f"Gene {gene} has GC content {gc*100:.1f}% and length {length} bp.")
```
